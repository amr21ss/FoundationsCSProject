```python
at_risk = students[(students['study_mode'] == 'Online') & (students['attendance_rate'] < 50) & (students['missed_deadlines'] > 5)] 
at_risk[['study_mode', 'attendance_rate', 'missed_deadlines']]
```

Question 2

```python
at_risk_by_course = at_risk.groupby('course_type').agg(students=('dropout', 'count'), dropout_rate=('dropout', 'mean')).sort_values(by='dropout_rate', ascending=False)
at_risk_by_course
```

Question 3

```python
students['pass_rate'] = students['units_passed'] / students['units_enrolled']
students[['pass_rate']]
```

```python
students['gpa_change'] = students['gpa_semester_2'] - students['gpa_semester_1']
def trend(row):
    if row['gpa_change'] > 0.5:
        return 'improving'
    elif row['gpa_change'] < -0.5:
        return 'declining'
    else:
        return 'stable'

students['gpa_trend'] = students.apply(trend, axis=1)
students[['gpa_change', 'gpa_trend']]
```
