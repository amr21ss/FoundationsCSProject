```python
at_risk = students[(students['study_mode'] == 'Online') & (students['attendance_rate'] < 50) & (students['missed_deadlines'] > 5)] 
at_risk[['study_mode', 'attendance_rate', 'missed_deadlines']]
```

```python
at_risk_by_course = at_risk.groupby('course_type').agg(students=('dropout', 'count'), dropout_rate=('dropout', 'mean')).sort_values(by='dropout_rate', ascending=False)
at_risk_by_course
```

```python
at_risk['gpa_change'] = at_risk['gpa_semester_2'] - at_risk['gpa_semester_1']
def trend(row):
    if row['gpa_change'] > 0.5:
        return 'improving'
    elif row['gpa_change'] < -0.5:
        return 'declining'
    else:
        return 'stable'

at_risk['gpa_trend'] = at_risk.apply(trend, axis=1)
at_risk[['gpa_change', 'gpa_trend']]
```
